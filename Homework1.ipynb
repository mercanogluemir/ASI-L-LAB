## Exercise 3 — Digit Extractor (algorithm building with `//` and `%`)

Remember the even/odd trick from Lesson 2, where `%` gave us the remainder? We can push that idea further.

Given `number = 68` (a two-digit number, given below — don't change it), figure out **without using `if`, loops, or string conversion** — only `//` and `%`:
- `tens` → the tens digit (hint: what do you get from `number // 10`?)
- `units` → the units digit (hint: what do you get from `number % 10`?)
- `digit_sum` → the sum of the two digits

This is exactly the kind of small building block real algorithms are made of — taking a number apart, digit by digit.

````python
# Exercise 3 -- write your code here
number = 68

# Define: tens, units, digit_sum (using only // and %)

tens = number // 10
units = number % 10
digit_sum = tens + units

print(f"Tens digit: {tens}")
print(f"Units digit: {units}")
print(f"Digit sum: {digit_sum}")
````

````python
check(tens == 6 and units == 8 and digit_sum == 14)
````

## Exercise 5 — Leap Year Checker (Boolean algorithm, no `if` needed)

A year is a leap year if:
- it is divisible by 4, **and**
- it is **not** divisible by 100, **unless** it is also divisible by 400

For example: 2000 is a leap year, 1900 is **not**, and 2024 **is**.

Using only comparison operators (`==`, `%`) and logical operators (`and`, `or`, `not`) — **no `if` statement** — compute a single `bool` for each year below and store the result:
- `is_leap_1900` for `year1 = 1900`
- `is_leap_2000` for `year2 = 2000`
- `is_leap_2023` for `year3 = 2023`

Hint: build one boolean expression, e.g. `(year % 4 == 0) and (...)`.

````python
# Exercise 5 -- write your code here
year1 = 1900
year2 = 2000
year3 = 2023 


# Define: is_leap_1900, is_leap_2000, is_leap_2023 (bool expressions, no if)


is_leap_1900 = (year1 % 4 == 0 and year1 % 100 != 0) or (year1 % 400 == 0)
is_leap_2000 = (year2 % 4 == 0 and year2 % 100 != 0) or (year2 % 400 == 0)
is_leap_2023 = (year3 % 4 == 0 and year3 % 100 != 0) or (year3 % 400 == 0)


print(is_leap_1900, is_leap_2000, is_leap_2023)
````

````python
check(is_leap_1900 == False and is_leap_2000 == True and is_leap_2023 == False)
````

## Exercise 8 — Final Challenge: Shopping Bill Calculator

This one combines *everything* above: arithmetic, comparisons, and a string method — without using `if`.

Given (don't change these):
```python
price = 250
quantity = 3
discount_code = "save10"
```

Steps:
1. `subtotal` → `price * quantity`
2. `discount_applied` → a **bool**: `True` if `discount_code`, made uppercase, equals `"SAVE10"`, otherwise `False`.
   (Research tip: you already know `.upper()` from Exercise 6 — use it here for a case-insensitive comparison.)
3. `discount_amount` → `subtotal * 0.10 * discount_applied`

   💡 **Algorithmic trick:** in Python, `True` behaves like `1` and `False` behaves like `0` in arithmetic. So multiplying by a `bool` turns it into an on/off switch — no `if` statement required! Try `print(True * 5)` somewhere to see it for yourself.
4. `total` → `subtotal - discount_amount`

Print `total`.

````python
# Exercise 8 -- write your code here
price = 250
quantity = 3
discount_code = "save10"

# Define: subtotal, discount_applied, discount_amount, total
subtotal = price * quantity
discount_applied = discount_code.upper() == "SAVE10"
discount_amount = subtotal * 0.1 * discount_applied
total = subtotal - discount_amount




print(f"Total: {total}")
````

````python
check(
    subtotal == 750
    and discount_applied is True
    and discount_amount == 75.0
    and total == 675.0
)
````
